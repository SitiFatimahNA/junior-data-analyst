# Initial Data Audit — News Category Dataset
Audit dataset dari Week 2 (Data Collection) - kategori JSON.

**Tujuan:** mengecek jumlah data, jumlah kolom, missing value, duplikat, tipe data, dan permasalahan lain sebelum dataset dipakai untuk analisis lanjutan.

```python
import pandas as pd

df = pd.read_json('News_Category_Dataset_v3.json', lines=True)
df.head()
```

## 1. Info Umum Dataset

```python
df.info()
```

## 2. Jumlah Data & Jumlah Kolom

```python
print('Jumlah baris :', df.shape[0])
print('Jumlah kolom :', df.shape[1])
```

## 3. Missing Value (Null Asli)

```python
df.isnull().sum()
```

## 4. Missing Value (String Kosong)

Data JSON hasil scraping sering menyimpan nilai kosong sebagai string `""`, bukan `null`, sehingga `isnull()` saja tidak cukup untuk mendeteksinya.

```python
text_cols = ['link', 'headline', 'category', 'short_description', 'authors']
for c in text_cols:
    n_empty = (df[c].astype(str).str.strip() == '').sum()
    print(f'{c}: {n_empty} baris kosong (empty string)')
```

## 5. Data Duplikat

```python
print('Duplikat (semua kolom identik):', df.duplicated().sum())
print('Duplikat berdasarkan link saja  :', df.duplicated(subset=["link"]).sum())
```

## 6. Tipe Data per Kolom

```python
df.dtypes
```

## 7. Statistik Deskriptif

```python
df.describe(include='all')
```

## 8. Pengecekan Kategori (Typo / Tumpang Tindih)

```python
print('Jumlah kategori unik:', df['category'].nunique())
print(sorted(df['category'].unique().tolist()))
```

```python
# Cek kelompok kategori yang kemungkinan bermakna sama
print('--- World News ---')
print(df[df['category'].str.contains('WORLD', case=False)]['category'].value_counts())
print()
print('--- Arts ---')
print(df[df['category'].str.contains('ARTS', case=False)]['category'].value_counts())
print()
print('--- Parenting ---')
print(df[df['category'].isin(['PARENTING', 'PARENTS'])]['category'].value_counts())
print()
print('--- Style ---')
print(df[df['category'].isin(['STYLE', 'STYLE & BEAUTY'])]['category'].value_counts())
```

## 9. Pengecekan Outlier Tanggal

```python
print('Rentang tanggal:', df['date'].min(), '-', df['date'].max())
outlier_date = ((df['date'] < '2010-01-01') | (df['date'] > '2023-01-01')).sum()
print('Jumlah baris dengan tanggal di luar rentang wajar (2010-2023):', outlier_date)
```

## 10. Contoh Baris Duplikat (Link)

```python
dupe_links = df[df.duplicated(subset=['link'], keep=False)].sort_values('link')
dupe_links[['link', 'headline', 'category']].head(6)
```
