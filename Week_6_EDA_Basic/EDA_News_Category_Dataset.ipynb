## 1. Apa itu EDA?

**Exploratory Data Analysis (EDA)** adalah proses menganalisis dan menyelidiki dataset untuk merangkum karakteristik utamanya, sering menggunakan visualisasi dan statistik deskriptif.

EDA dilakukan **sebelum** membangun model atau menarik kesimpulan, dengan tujuan untuk:
- Memahami struktur dan bentuk data
- Menemukan pola, anomali, atau outlier
- Menguji asumsi awal terhadap data
- Menentukan langkah analisis/pemodelan selanjutnya yang tepat

> EDA menjawab pertanyaan: **"Data seperti apa yang sedang saya hadapi?"** sebelum menjawab "Apa insight/model yang bisa dibangun dari data ini?"


## 2. Tujuan EDA

- Memahami distribusi tiap variabel (numerik maupun kategorikal)
- Mendeteksi missing value, duplicate, atau data tidak konsisten yang mungkin masih tersisa
- Melihat hubungan antar variabel (korelasi, pola)
- Mengidentifikasi outlier yang dapat mempengaruhi analisis
- Menghasilkan hipotesis awal untuk analisis lebih lanjut


## 3. Workflow EDA

1. **Load data** — baca dataset ke dalam DataFrame
2. **Overview** — cek ukuran, tipe data, dan struktur kolom (`.info()`, `.head()`)
3. **Statistik deskriptif** — `.describe()`, `.mean()`, `.median()`, dll.
4. **Analisis univariate** — eksplorasi tiap kolom secara individual
5. **Analisis bivariate/multivariate** — eksplorasi hubungan antar kolom
6. **Visualisasi** — histogram, boxplot, bar chart, dsb.
7. **Insight & dokumentasi** — rangkum temuan awal


## 4. Jenis Analisis dalam EDA

| Jenis | Definisi | Contoh pada dataset ini |
|---|---|---|
| **Univariate** | Analisis satu variabel secara terpisah | Distribusi `Category`, rata-rata `Headline_Word_Count` |
| **Bivariate** | Analisis hubungan antara dua variabel | `Category` vs `Headline_Word_Count`, `Year` vs jumlah artikel |
| **Multivariate** | Analisis hubungan antara tiga variabel atau lebih | `Category`, `Author_Type`, dan `Year` sekaligus |

## Tugas: Analisis Statistik Dasar Dataset

Menghitung **Mean, Median, Modus, Minimum, Maksimum, Standar Deviasi, dan Kuartil** untuk kolom numerik, serta statistik ringkasan untuk kolom kategorikal.

### Identifikasi kolom numerik

```python
numeric_cols = df.select_dtypes(include='number').columns.tolist()
numeric_cols
```

### Statistik dasar — kolom numerik

```python
stats_rows = []
for col in numeric_cols:
    stats_rows.append({
        'Column': col,
        'Mean': round(df[col].mean(), 2),
        'Median': df[col].median(),
        'Modus': df[col].mode()[0],
        'Min': df[col].min(),
        'Max': df[col].max(),
        'Std_Dev': round(df[col].std(), 2),
        'Q1 (25%)': df[col].quantile(0.25),
        'Q2 (50%)': df[col].quantile(0.50),
        'Q3 (75%)': df[col].quantile(0.75),
        'IQR': round(df[col].quantile(0.75) - df[col].quantile(0.25), 2),
    })

numeric_summary = pd.DataFrame(stats_rows)
numeric_summary
```

### Interpretasi singkat

- **Year**: median 2015 — separuh artikel diterbitkan sebelum 2015, separuh sesudahnya. Rentang tahun 2012–2022.
- **Month**: distribusi cukup merata sepanjang bulan (mean ≈ median ≈ 6).
- **Headline_Word_Count**: rata-rata judul terdiri dari ±9-10 kata, dengan sebaran (std) sekitar 3 kata — cukup konsisten.
- **Num_Authors**: modus 1 penulis per artikel, tapi ada artikel dengan lebih dari 1 penulis (kolaboratif).

### Statistik dasar — kolom kategorikal (modus & frekuensi)

```python
cat_cols = ['Category', 'Day_Of_Week', 'Author_Type']

cat_rows = []
for col in cat_cols:
    vc = df[col].value_counts()
    cat_rows.append({
        'Column': col,
        'Unique_Values': df[col].nunique(),
        'Modus (Paling Sering)': vc.index[0],
        'Modus_Count': int(vc.iloc[0]),
        'Modus_Percentage (%)': round(vc.iloc[0] / len(df) * 100, 2),
    })

categorical_summary = pd.DataFrame(cat_rows)
categorical_summary
```

### Ringkasan lengkap `describe()` untuk seluruh kolom (numerik + kategorikal)

```python
full_describe = df.describe(include='all').transpose()
full_describe
```

## Output: Export ke `Statistical Summary.xlsx`

Menyimpan seluruh hasil analisis statistik ke dalam satu file Excel dengan beberapa sheet.

```python
with pd.ExcelWriter('Statistical Summary.xlsx', engine='openpyxl') as writer:
    numeric_summary.to_excel(writer, sheet_name='Numeric Summary', index=False)
    categorical_summary.to_excel(writer, sheet_name='Categorical Summary', index=False)
    full_describe.to_excel(writer, sheet_name='Describe Full')
    df['Category'].value_counts().rename_axis('Category').reset_index(name='Count').to_excel(writer, sheet_name='Category Value Counts', index=False)
    df['Day_Of_Week'].value_counts().rename_axis('Day_Of_Week').reset_index(name='Count').to_excel(writer, sheet_name='Day of Week Counts', index=False)

print("File 'Statistical Summary.xlsx' berhasil dibuat.")
```
